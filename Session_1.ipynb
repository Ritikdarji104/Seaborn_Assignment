```python
# task 2

# Weekly Temperature Trend using sns.lineplot()

import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.show()
```

```python
# tsk 3

# Axes-level vs Figure-level
# A. Axes-level function — sns.lineplot()

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.show()
```

```python
# Task 3

# B. Figure-level function — sns.relplot()

sns.set_theme(style="whitegrid")

sns.relplot(
    x=days,
    y=temps,
    kind="line",
    marker="o"
)

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.show()
```

```python
# task 4

# 4. Seaborn Themes + Save Images
# Theme 1 — Dark Grid

sns.set_theme(style="darkgrid")

plt.figure(figsize=(8, 5))

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.savefig("temperature_darkgrid.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# task 4

# Theme 2 — White Grid

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.savefig("temperature_whitegrid.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# task 4

# Theme 3 — Ticks

sns.set_theme(style="ticks")

plt.figure(figsize=(8, 5))

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.savefig("temperature_ticks.png", dpi=150, bbox_inches="tight")
plt.show()
```
