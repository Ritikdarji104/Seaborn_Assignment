```python
# Task 1

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample Instagram post data
data = {
    'likes': [120, 250, 400, 550, 700, 900, 1100, 1350, 1600, 2000, 2400, 2800],
    'comments': [10, 18, 25, 32, 40, 48, 55, 65, 72, 85, 95, 110]
}

instagram_df = pd.DataFrame(data)

# Regression plot
sns.regplot(
    data=instagram_df,
    x='likes',
    y='comments'
)

plt.title('Instagram Likes vs Comments')
plt.xlabel('Number of Likes')
plt.ylabel('Number of Comments')
plt.show()
```

```python
# Task 2

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import random

# Sample Zomato-style dataset
data = {
    'order_price': [
        120, 250, 180, 350, 500,
        220, 450, 300, 150, 600,
        275, 400, 175, 525, 325
    ],
    'delivery_time': [
        25, 35, 28, 42, 50,
        32, 45, 38, 22, 55,
        34, 44, 27, 52, 40
    ]
}

zomato_df = pd.DataFrame(data)

# Regression plot
sns.lmplot(
    data=zomato_df,
    x='order_price',
    y='delivery_time'
)

plt.title('Food Order Price vs Delivery Time')
plt.xlabel('Order Price (₹)')
plt.ylabel('Delivery Time (Minutes)')
plt.show()
```

```python
# Task 3

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

import statsmodels

print(statsmodels.__version__)

# Sample Spotify playlist data
data = {
    'songs': [
        10, 15, 20, 25, 30,
        35, 40, 45, 50, 55,
        60, 65, 70, 75, 80
    ],
    'duration': [
        35, 52, 70, 88, 105,
        125, 140, 160, 175, 200,
        218, 230, 255, 270, 290
    ]
}

spotify_df = pd.DataFrame(data)

# LOWESS regression plot
sns.regplot(
    data=spotify_df,
    x='songs',
    y='duration',
    lowess=True
)

plt.title('Number of Songs vs Playlist Duration')
plt.xlabel('Number of Songs')
plt.ylabel('Playlist Duration (Minutes)')
plt.show()
```

```python
# Task 5

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample Flipkart product data
data = {
    'price': [
        499, 799, 999, 1299, 1599,
        1999, 2499, 2999, 3499, 3999,
        4499, 4999, 5999, 6999, 7999
    ],
    'reviews': [
        850, 720, 680, 600, 550,
        500, 450, 420, 390, 350,
        320, 300, 270, 240, 210
    ]
}

flipkart_df = pd.DataFrame(data)

# Original regression logic remains the same
plot = sns.lmplot(
    data=flipkart_df,
    x='price',
    y='reviews',
    scatter_kws={'color': 'purple'}
)

# Plot aesthetics
plot.set_axis_labels('Product Price (₹)', 'Number of Reviews')
plot.fig.suptitle(
    'Flipkart Product Price vs Number of Reviews',
    y=1.02
)

plt.show()
```
