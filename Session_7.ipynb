```python
# Task 1

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Random seed for reproducible results
np.random.seed(42)

# Create Food Delivery dataset
data = {
    'delivery_time': np.random.randint(20, 61, 20),
    'order_value': np.random.randint(150, 1001, 20),
    'tip_amount': np.random.randint(0, 101, 20),
    'rating': np.round(np.random.uniform(2.5, 5.0, 20), 1),
    'delivery_distance': np.round(np.random.uniform(1, 15, 20), 2)
}

df = pd.DataFrame(data)

print("Food Delivery Dataset:")
display(df)

# Calculate correlation matrix
correlation_matrix = df.corr()

print("\nCorrelation Matrix:")
display(correlation_matrix)
```

```python
# Task 2

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Food Delivery Features Correlation Heatmap')
plt.show()
```

```python
# Task 3

# Copy correlation matrix
corr = correlation_matrix.copy()

# Remove diagonal values
for i in range(len(corr)):
    corr.iloc[i, i] = np.nan

# Convert correlation matrix into pairs
corr_pairs = corr.unstack().dropna()

# Find strongest negative correlation
strongest_negative = corr_pairs[corr_pairs < 0].idxmin()
strongest_negative_value = corr_pairs[corr_pairs < 0].min()

print("Strongest Negative Correlation:")
print(f"Features: {strongest_negative[0]} and {strongest_negative[1]}")
print(f"Correlation: {strongest_negative_value:.2f}")
```

```python
# Task 4

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create mobile phone dataset
data = {
    'price': [12000, 15000, 18000, 22000, 25000,
              28000, 32000, 35000, 40000, 45000,
              50000, 55000, 60000, 65000, 70000],

    'rating': [3.8, 4.0, 4.1, 4.2, 4.0,
               4.3, 4.4, 4.2, 4.5, 4.3,
               4.6, 4.5, 4.7, 4.6, 4.8],

    'ram': [4, 4, 6, 6, 8,
            8, 8, 12, 12, 12,
            12, 16, 16, 16, 16],

    'battery_capacity': [4000, 4500, 4500, 5000, 5000,
                         5000, 5000, 5000, 5200, 5200,
                         5500, 5500, 6000, 6000, 6000]
}

# Create DataFrame
mobile_df = pd.DataFrame(data)

# Display dataset
print("Mobile Phones Dataset:")
display(mobile_df)

# Calculate correlation matrix
mobile_corr = mobile_df.corr()

print("Mobile Phones Correlation Matrix:")
display(mobile_corr)

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    mobile_corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Mobile Phones Features Correlation Heatmap')
plt.show()
```
