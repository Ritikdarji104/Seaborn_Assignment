```python
# Task 1
# Que--1-- Use sns.countplot() to visualize the number of orders placed by each payment method (e.g., UPI, Card, Cash) in a sample Swiggy orders DataFrame with at least 3 payment types.


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Payment_Method': ['UPI', 'Card', 'Cash', 'UPI', 'Card',
                       'UPI', 'Cash', 'UPI', 'Card', 'Cash']
}

df = pd.DataFrame(data)

sns.countplot(x='Payment_Method', data=df)

plt.title('Orders by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.show()
```

```python
# Task 2
# Create a barplot using sns.barplot() to show the average delivery time (in minutes) for each food category (e.g., Pizza, Burger, Biryani, South Indian) from a Zomato-style dataset.<br><br><em><strong>Hint:</strong> Use the estimator parameter to show the mean delivery time for each category.</em>


data1 = {
    'Food_Category': ['Pizza', 'Burger', 'Biryani', 'South Indian',
                      'Pizza', 'Burger', 'Biryani', 'South Indian',
                      'Pizza', 'Burger'],
    
    'Delivery_Time': [35, 25, 40, 30, 45, 20, 50, 35, 30, 25]
}

df = pd.DataFrame(data1)

sns.barplot(
    x='Food_Category',
    y='Delivery_Time',
    data=df,
    estimator='mean'
)

plt.title('Average Delivery Time by Food Category')
plt.xlabel('Food Category')
plt.ylabel('Average Delivery Time (Minutes)')
plt.show()
```

```python
# Task 3

# Modify your barplot to display the total number of orders instead of the average by setting estimator to sum in sns.barplot().

data2 = {
    'Food_Category': ['Pizza', 'Burger', 'Biryani', 'South Indian',
                      'Pizza', 'Burger', 'Biryani', 'South Indian',
                      'Pizza', 'Burger'],
    
    'Orders': [10, 15, 8, 12, 20, 18, 14, 10, 25, 22]
}

df = pd.DataFrame(data2)

sns.barplot(
    x='Food_Category',
    y='Orders',
    data=df,
    estimator='sum'
)

plt.title('Total Orders by Food Category')
plt.xlabel('Food Category')
plt.ylabel('Total Orders')
plt.show()
```
