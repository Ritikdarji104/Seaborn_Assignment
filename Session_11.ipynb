```python
# Task 1
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample food order data
data = {
    'category': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'orders': [450, 380, 520, 300, 250]
}

food_df = pd.DataFrame(data)

# Create barplot
plt.figure(figsize=(8, 5))

sns.barplot(
    data=food_df,
    x='category',
    y='orders'
)

plt.title('Number of Orders by Food Category')
plt.xlabel('Food Category')
plt.ylabel('Number of Orders')

# Save as PNG with 300 DPI
plt.savefig(
    'food_orders.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
# Task 2

import seaborn as sns
import matplotlib.pyplot as plt

# One week step counts
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
steps = [6500, 7200, 5800, 8000, 7500, 9500, 8800]

plt.figure(figsize=(9, 5))

# Lineplot
sns.lineplot(
    x=days,
    y=steps,
    marker='o'
)

plt.title('Daily Step Counts - One Week')
plt.xlabel('Day')
plt.ylabel('Number of Steps')

# Export as SVG
plt.savefig(
    'weekly_steps.svg',
    format='svg',
    bbox_inches='tight'
)

plt.show()
```

```python
# Task 3

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Apply darkgrid style
sns.set_style('darkgrid')

# Sample sales data
data = {
    'category': ['Mobiles', 'Laptops', 'TVs', 'Headphones', 'Accessories'],
    'sales': [85000, 72000, 55000, 38000, 46000]
}

sales_df = pd.DataFrame(data)

plt.figure(figsize=(9, 6))

# Barplot with custom palette
sns.barplot(
    data=sales_df,
    x='category',
    y='sales',
    palette='muted'
)

plt.title('Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Sales (₹)')

# Save high-resolution PNG
plt.savefig(
    'sales_chart_1200dpi.png',
    dpi=1200,
    bbox_inches='tight'
)

plt.show()
```

```python
# Task 4

import seaborn as sns
import matplotlib.pyplot as plt

# Presentation theme function
def apply_presentation_theme():
    sns.set_theme(
        context='talk',
        style='whitegrid',
        palette='deep',
        font_scale=1.1
    )


# Apply presentation theme
apply_presentation_theme()


# Top 5 most-used apps
apps = [
    'Instagram',
    'YouTube',
    'WhatsApp',
    'Spotify',
    'Chrome'
]

usage = [120, 105, 95, 70, 60]

# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    usage,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Most-Used Apps')

# Export chart
plt.savefig(
    'top_5_apps.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
# Task 5

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# ---------------------------------------------------------
# ChatGPT Prompt:
#
# "Generate a code snippet for exporting a Seaborn heatmap
# as a 1200 DPI SVG file."
# ---------------------------------------------------------
#
# Code received from ChatGPT:
#
# sns.heatmap(data)
# plt.savefig('heatmap.svg', format='svg', dpi=1200)
# ---------------------------------------------------------


# My own heatmap data
data = {
    'Math': [85, 90, 78, 92, 88],
    'Science': [80, 85, 82, 95, 90],
    'English': [75, 88, 80, 85, 92],
    'Computer': [95, 92, 98, 90, 96],
    'History': [70, 82, 78, 88, 85]
}

df = pd.DataFrame(data)

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    df,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title('Student Subject Scores')
plt.xlabel('Subjects')
plt.ylabel('Students')

# Save as SVG
plt.savefig(
    'student_heatmap.svg',
    format='svg',
    dpi=1200,
    bbox_inches='tight'
)

plt.show()
```
