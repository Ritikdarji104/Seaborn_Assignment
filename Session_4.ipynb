```python
# Task 1
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

prices = [499, 799, 999, 1299, 1499, 1599, 1799, 1999,
          2299, 2499, 2799, 2999, 3499, 3999, 4500, 12000]

df = pd.DataFrame({"Price": prices})

sns.boxplot(x=df["Price"])

plt.title("Flipkart Product Price Distribution")
plt.xlabel("Price (₹)")
plt.show()
```

```python
# Task 2
data = {
    "Category": [
        "Fast Food", "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Fast Food", "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Desserts", "Desserts", "Desserts", "Desserts", "Desserts",
        "Desserts", "Desserts", "Desserts", "Desserts", "Desserts"
    ],
    "Order_Amount": [
        180, 220, 250, 300, 350,
        280, 400, 450, 320, 270,
        120, 150, 180, 200, 250,
        300, 220, 170, 280, 350
    ]
}

df = pd.DataFrame(data)

sns.violinplot(
    data=df,
    x="Category",
    y="Order_Amount"
)

plt.title("Zomato Order Amount by Restaurant Category")
plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")
plt.show()
```

```python
# Task 3


ticket_prices = [
    500, 600, 700, 800, 900,
    1000, 1100, 1200, 1300, 1400,
    1500, 1600, 1700, 1800, 2000,
    2200, 2500, 3000, 8000
]

sns.boxplot(x=ticket_prices)

plt.title("IPL Match Ticket Price Distribution")
plt.xlabel("Ticket Price (₹)")
plt.show()
```

```python
# for outliers 

import numpy as np

Q1 = np.percentile(ticket_prices, 25)
Q3 = np.percentile(ticket_prices, 75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = [
    price for price in ticket_prices
    if price < lower_limit or price > upper_limit
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:", outliers)
```
