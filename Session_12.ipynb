```python
# Task 1

import pandas as pd

# Load dataset
bank_df = pd.read_csv(r"C:\Users\Dhaval Darji\Downloads\bank-additional-full.csv", sep=';')

# Display first 10 rows
print("First 10 rows:")
display(bank_df.head(10))

# Display DataFrame information
print("\nDataFrame Information:")
bank_df.info()
```

```python
bank_df.shape
```

```python
# Task 2

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(
    data=bank_df,
    x='y'
)

plt.title('Term Deposit Subscription')
plt.xlabel('Subscribed to Term Deposit')
plt.ylabel('Number of Customers')

plt.show()
```

```python
# Task 3

plt.figure(figsize=(8, 6))

sns.boxplot(
    data=bank_df,
    x='y',
    y='age'
)

plt.title('Customer Age vs Term Deposit Subscription')
plt.xlabel('Term Deposit Subscription')
plt.ylabel('Customer Age')

plt.show()
```

```python
# Task 4

# Calculate subscription percentage by education
education_percentage = (
    bank_df.groupby('education')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .reset_index(name='subscription_percentage')
)

# Display calculated percentages
display(education_percentage)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=education_percentage,
    x='education',
    y='subscription_percentage'
)

plt.title('Term Deposit Subscription Percentage by Education')
plt.xlabel('Education Level')
plt.ylabel('Subscription Percentage (%)')

plt.xticks(rotation=30)
plt.show()
```

```python
# Task 5

# Calculate subscription percentage by job
job_subscription = (
    bank_df.groupby('job')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .reset_index(name='subscription_percentage')
)

# Sort from highest to lowest
job_subscription = job_subscription.sort_values(
    'subscription_percentage',
    ascending=False
)

display(job_subscription)
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=job_subscription,
    x='job',
    y='subscription_percentage'
)

plt.title('Term Deposit Subscription Rate by Job')
plt.xlabel('Job Type')
plt.ylabel('Subscription Percentage (%)')

plt.xticks(rotation=45)
plt.show()
```
