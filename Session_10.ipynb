```python
# Task 1

import seaborn as sns
import matplotlib.pyplot as plt

# Spotify streaming data
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
songs_streamed = [45, 60, 52, 70, 85, 110, 95]

# Apply poster context
sns.set(context='poster')

# Barplot
sns.barplot(
    x=days,
    y=songs_streamed
)

plt.title('Spotify Songs Streamed Per Day')
plt.xlabel('Day')
plt.ylabel('Number of Songs')
plt.show()
```

```python
# Task 2

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample food order data
data = {
    'cuisine': [
        'Indian', 'Chinese', 'Italian', 'Indian', 'Chinese',
        'Indian', 'Italian', 'Chinese', 'Indian', 'Italian',
        'Chinese', 'Indian', 'Italian', 'Chinese', 'Indian',
        'Italian', 'Indian', 'Chinese', 'Italian', 'Indian'
    ]
}

food_df = pd.DataFrame(data)

# Countplot with colorblind palette
sns.countplot(
    data=food_df,
    x='cuisine',
    palette='colorblind'
)

plt.title('Food Orders by Cuisine Type')
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')
plt.show()
```

```python
# Task 3

import seaborn as sns
import matplotlib.pyplot as plt
import random
import pandas as pd

# 14 days
days = [
    'Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5',
    'Day 6', 'Day 7', 'Day 8', 'Day 9', 'Day 10',
    'Day 11', 'Day 12', 'Day 13', 'Day 14'
]

# Simulate step counts
steps = [random.randint(4000, 12000) for _ in range(14)]

steps_df = pd.DataFrame({
    'day': days,
    'steps': steps
})

# Apply notebook context and muted palette
sns.set(
    context='notebook',
    palette='muted'
)

# Lineplot
sns.lineplot(
    data=steps_df,
    x='day',
    y='steps',
    marker='o'
)

plt.title('Daily Step Counts - 2 Weeks')
plt.xlabel('Day')
plt.ylabel('Number of Steps')
plt.xticks(rotation=45)
plt.show()
```

```python
# Task 5

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# -----------------------------
# Flipkart-style Seaborn Theme
# -----------------------------

sns.set(
    context='talk',
    style='whitegrid',
    palette='deep',
    font_scale=1.1,
    rc={
        'grid.linestyle': '--',
        'grid.linewidth': 0.8
    }
)

# Sales data
data = {
    'category': [
        'Mobiles',
        'Laptops',
        'TVs',
        'Headphones',
        'Accessories'
    ],
    'sales': [
        85000,
        72000,
        55000,
        38000,
        46000
    ]
}

sales_df = pd.DataFrame(data)

# Barplot
sns.barplot(
    data=sales_df,
    x='category',
    y='sales'
)

plt.title('Flipkart Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Sales (₹)')
plt.xticks(rotation=20)

plt.show()
```
