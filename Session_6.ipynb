```python
# Task 1

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "rating": [3.2, 3.5, 3.8, 4.0, 4.1, 4.2, 4.3, 4.4, 4.5, 4.6,
               4.7, 4.8, 3.9, 4.2, 4.5],
    "price": [499, 799, 999, 1299, 1599, 1899, 2299, 2499,
              2999, 3499, 3999, 4999, 1199, 1999, 2999]
}

df = pd.DataFrame(data)

sns.jointplot(
    x="rating",
    y="price",
    data=df,
    kind="scatter"
)

plt.show()
```

```python
# Task 4

data = {
    "rating": [
        3.0, 3.1, 3.2, 3.3, 3.4,
        3.5, 3.6, 3.7, 3.8, 3.9,
        4.0, 4.1, 4.2, 4.3, 4.4,
        4.5, 4.6, 4.7, 4.8, 4.9
    ],
    "reviews": [
        50, 80, 120, 150, 200,
        250, 300, 350, 450, 500,
        600, 700, 850, 1000, 1200,
        1500, 1800, 2200, 3000, 5000
    ]
}

df = pd.DataFrame(data)
```
