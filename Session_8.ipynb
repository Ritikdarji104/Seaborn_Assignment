```python
# Task 1

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Fictional Instagram users
data = {
    'user': [
        'User1', 'User2', 'User3', 'User4', 'User5',
        'User6', 'User7', 'User8', 'User9', 'User10',
        'User11', 'User12', 'User13', 'User14', 'User15',
        'User16', 'User17', 'User18', 'User19', 'User20'
    ],
    'followers': [
        1200, 2500, 5000, 3200, 8500,
        15000, 7200, 12000, 18000, 25000,
        30000, 45000, 38000, 52000, 60000,
        75000, 90000, 110000, 135000, 160000
    ],
    'posts': [
        45, 80, 120, 65, 150,
        210, 180, 250, 300, 350,
        420, 500, 450, 600, 700,
        850, 900, 1100, 1250, 1400
    ]
}

instagram_df = pd.DataFrame(data)

# Scatter plot
sns.scatterplot(
    data=instagram_df,
    x='followers',
    y='posts'
)

plt.title('Instagram Followers vs Number of Posts')
plt.xlabel('Number of Followers')
plt.ylabel('Number of Posts')
plt.show()
```

```python
# Task 2

import seaborn as sns
import matplotlib.pyplot as plt
import random

# Days of the month
days = list(range(1, 31))

# Generate 30 random order counts
orders = [random.randint(100, 500) for _ in range(30)]

# Line plot
sns.lineplot(
    x=days,
    y=orders,
    marker='o'
)

plt.title('Daily Zomato Food Orders - One Month')
plt.xlabel('Day of the Month')
plt.ylabel('Number of Food Orders')
plt.xticks(days)
plt.show()
```

```python
# Task 3

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample movie dataset
data = {
    'movie': [
        'Movie1', 'Movie2', 'Movie3', 'Movie4', 'Movie5',
        'Movie6', 'Movie7', 'Movie8', 'Movie9', 'Movie10',
        'Movie11', 'Movie12', 'Movie13', 'Movie14', 'Movie15'
    ],
    'rating': [
        8.2, 7.5, 8.8, 6.9, 7.8,
        8.5, 7.2, 9.0, 6.8, 8.1,
        7.7, 8.9, 7.0, 8.4, 7.6
    ],
    'reviews': [
        1200, 850, 2300, 600, 1500,
        1900, 700, 3000, 500, 1800,
        1100, 2700, 650, 2100, 950
    ],
    'genre': [
        'Action', 'Comedy', 'Drama', 'Action', 'Comedy',
        'Drama', 'Action', 'Drama', 'Comedy', 'Action',
        'Comedy', 'Drama', 'Action', 'Drama', 'Comedy'
    ]
}

movies_df = pd.DataFrame(data)

# Relational plot
sns.relplot(
    data=movies_df,
    x='rating',
    y='reviews',
    hue='genre',
    kind='scatter'
)

plt.title('Movie Ratings vs Number of Reviews')
plt.xlabel('Movie Rating')
plt.ylabel('Number of Reviews')
plt.show()
```

```python
# Task 4

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Data for 25 students
data = {
    'youtube_hours': [
        1, 2, 3, 1.5, 4,
        2.5, 5, 3.5, 6, 2,
        7, 4.5, 8, 3, 5.5,
        6.5, 2.5, 7.5, 4, 8.5,
        3.5, 9, 5, 6, 10
    ],
    'exam_score': [
        92, 88, 84, 90, 78,
        85, 72, 80, 68, 87,
        60, 75, 55, 83, 70,
        65, 86, 58, 77, 52,
        81, 50, 73, 67, 45
    ]
}

students_df = pd.DataFrame(data)

# Regression plot
sns.regplot(
    data=students_df,
    x='youtube_hours',
    y='exam_score'
)

plt.title('YouTube Hours vs Exam Scores')
plt.xlabel('Hours Spent on YouTube')
plt.ylabel('Exam Score')
plt.show()
```
